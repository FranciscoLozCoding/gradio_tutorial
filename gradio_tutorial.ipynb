# Tutorial: Hands-on with Gradio

This tutorial will implement Gradio with [MobileNetV2](https://arxiv.org/abs/1801.04381), to show how Gradio can be used with ML for an easy way to demo your models.

## Import Libraries

First we will import libraries that we will need in this tutorial.

```python
import gradio as gr
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input, decode_predictions
from PIL import Image
import numpy as np
import gradio as gr
```

## Create a Basic Interface

First we will create a "hello world" app to see how Gradio works.

```python
def greet(name):
    return f"Hello, {name}!"

demo = gr.Interface(fn=greet, inputs="text", outputs="text")
demo.launch()
```

## Implement a Machine Learning Model

Now that we know how Gradio works, we will implement Gradio for creating a MobileNetV2 demo.

```python
model = MobileNetV2(weights="imagenet")

def classify_image(img):
    img = img.resize((224, 224))
    img = np.array(img)
    img = preprocess_input(img)
    img = np.expand_dims(img, axis=0)
    preds = model.predict(img)
    label = decode_predictions(preds, top=1)[0][0][1]
    return label

demo = gr.Interface(fn=classify_image, inputs="image", outputs="text")
demo.launch()
```

## Deploy and Share

Gradio can also be used to share your ML-powered demos. You can also deploy to [HuggingFace Spaces](https://huggingface.co/docs/hub/en/spaces-overview). HuggingFace Spaces is a place to host your demos. Here is an example of an app being hosted in HF spaces, [AI Comic Factory](https://huggingface.co/spaces/jbilcke-hf/ai-comic-factory)

```python
demo.launch(share=True)
```

## Customizing UI

Here I will show how to customize your UI using Gradio.

```python
demo = gr.Interface(
    fn=classify_image,
    inputs=gr.Image(shape=(224, 224)),
    outputs=gr.Textbox(label="Prediction"),
    title="Image Classifier",
    description="Upload an image to classify it using MobileNetV2."
)
demo.launch()
```
